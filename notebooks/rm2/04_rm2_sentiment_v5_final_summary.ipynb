# RM2 Sentiment IndoBERT V5 Final Summary

This notebook documents the accepted canonical V5 sentiment outputs. It does not train a model, open locked test data, or rerun inference.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().parents[1] if Path.cwd().name == 'rm2' else Path.cwd()
FINAL = ROOT / 'output/rm2_sentiment/final'
canonical = json.loads((FINAL / 'CANONICAL_MODEL.json').read_text(encoding='utf-8'))
manifest = json.loads((FINAL / 'INDOBERT_V5_FULL_INFERENCE_MANIFEST.json').read_text(encoding='utf-8'))
print(canonical['canonical_status'])
print('observational rows:', manifest['observational_rows'])
```

```python
dist = pd.read_csv(FINAL / 'tables/sentiment_distribution_observational_v5.csv')
hcc_goals = pd.read_csv(FINAL / 'tables/hcc_sentiment_goals_summary_v5.csv')
actor_goals = pd.read_csv(FINAL / 'tables/actor_type_goals_pooled_v5.csv')
display(dist)
display(hcc_goals[['hcc_id', 'goal_orientation', 'dominant_sentiment', 'evaluable_comments']].head())
display(actor_goals)
```

Goal orientation is a descriptive aggregate of observed message sentiment. It is not evidence of intent, payment, buzzer status, affiliation, or causal influence. RM1 topology, LCN, HCC membership, nodes, edges, and community membership are not recalculated here.
